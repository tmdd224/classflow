## **Normalisation des données**

```python
from sklearn.preprocessing import MinMaxScaler

# 1. On instancie le scaler
scaler = MinMaxScaler()

# 2. Liste des colonnes numériques à normaliser
cols_numeriques = ['temp_max_c', 'temp_min_c', 'humidite_pct', 'pluie_mm', 'vent_kmh']

# 3. Création d'une copie du DataFrame et application de la normalisation
df_scaled = df.copy()
df_scaled[cols_numeriques] = scaler.fit_transform(df[cols_numeriques])

# 4. Affichage du résultat
df_scaled.head(10)
```

## **Comparaison boucle Python et NumPy avec %timeit**

```python
# 1. On crée une liste de 1 million de chiffres
data_list = list(range(100000))
data_array = np.array(data_list)

# 2. Temps avec la fonction Python classique
%timeit max(data_list)

# 3. Temps avec la fonction optimisée NumPy
%timeit data_array.max()
```
