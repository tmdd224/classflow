3. Moyennes, extrêmes et jours de canicule par mois
Objectif
À partir des températures de Dakar, nous devons obtenir pour chaque mois :
- la température moyenne ;
- la température minimale ;
- la température maximale ;
- le nombre de jours de canicule.

```python
stats_mensuelles = df.groupby("mois").agg(
    temp_max_moyenne=("temp_max_c", "mean"),
    temp_max_min=("temp_max_c", "min"),
    temp_max_max=("temp_max_c", "max"),
    temp_min_moyenne=("temp_min_c", "mean"),
    temp_min_min=("temp_min_c", "min"),
    temp_min_max=("temp_min_c", "max")
)

stats_mensuelles
```

### Interprétation des statistiques mensuelles

L'analyse des températures par mois permet d'identifier les périodes les plus chaudes et les plus fraîches de l'année à partir des valeurs réellement observées dans les données.

- La **température maximale moyenne** permet d'identifier les mois présentant les journées généralement les plus chaudes.
- La **température minimale moyenne** permet de comparer les températures les plus basses observées habituellement selon les mois.
- Les valeurs **minimales et maximales absolues** permettent d'identifier les extrêmes enregistrés au cours de chaque mois.
- Le mois présentant la température maximale moyenne la plus élevée correspond à la période globalement la plus chaude de l'année dans le jeu de données.
- À l'inverse, le mois présentant la température minimale moyenne la plus faible correspond à la période globalement la plus fraîche.

L'interprétation est basée sur les valeurs calculées à partir des données de Dakar. Les notions de forte chaleur ou de canicule seront déterminées selon le seuil défini dans le TP.

4.1 Normalise temp_max_c

```python
temp_max_min = df["temp_max_c"].min()
temp_max_max = df["temp_max_c"].max()

df["temp_max_normalisee"] = (
    (df["temp_max_c"] - temp_max_min)
    / (temp_max_max - temp_max_min)
)

df[["temp_max_c", "temp_max_normalisee"]].head()
```

Vérifie que la normalisation est correcte

```python
df["temp_max_normalisee"].min(), df["temp_max_normalisee"].max()
```


La normalisation Min-Max transforme les températures maximales sur une échelle comprise entre 0 et 1.

La valeur 0 correspond à la température maximale la plus faible observée dans le jeu de données, tandis que la valeur 1 correspond à la température maximale la plus élevée.

Cette transformation conserve l'ordre relatif des températures tout en les ramenant sur une échelle commune.

5. Comparer une boucle Python et NumPy avec %timeit

On récupère les températures maximales sous forme de tableau NumPy 

```python
temperatures = df["temp_max_c"].to_numpy()
```

 calcul le carré de chaque température avec une boucle

```python
def calcul_boucle(data):
    resultat = []
    for x in data:
        resultat.append(x ** 2)
    return resultat
calcul_boucle(temperatures)[:5]
```

```python
def calcul_numpy(data):
    return data ** 2
calcul_numpy(temperatures)[:5]
```

5.4 Comparaison avec %timeit

```python
%timeit calcul_boucle(temperatures)
```

```python
%timeit calcul_numpy(temperatures)
```

```python
np.allclose(
    calcul_boucle(temperatures),
    calcul_numpy(temperatures)
)
```



Les deux méthodes réalisent le même calcul, ce qui est vérifié par `np.allclose()`.

La méthode utilisant une boucle Python traite les températures une par une, tandis que NumPy réalise le calcul de manière vectorisée sur l'ensemble du tableau.

Les mesures obtenues avec `%timeit` montrent que la méthode la plus rapide est celle qui présente le temps d'exécution moyen le plus faible. Cette comparaison met ainsi en évidence l'intérêt de la vectorisation NumPy pour effectuer efficacement des opérations numériques sur des données.

La boucle Python est utilisée ici uniquement comme méthode de référence afin de comparer ses performances avec celles de NumPy, conformément à l'objectif du TP.
